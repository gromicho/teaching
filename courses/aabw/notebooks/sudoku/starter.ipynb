```python
import pandas as pd
from io import StringIO
```

```python
def print_sudoku(df):
    def psv(r, v, df=df):
        if df.loc[r, v] == 0:
            return '_'
        return str(df.loc[r, v])

    for r in df.index:
        print(psv(r, 1) + ' ' + psv(r, 2) + ' ' + psv(r, 3) + ' | ' +
              psv(r, 4) + ' ' + psv(r, 5) + ' ' + psv(r, 6) + ' | ' +
              psv(r, 7) + ' ' + psv(r, 8) + ' ' + psv(r, 9))
        if r % 3 == 0 and r != 9:
            print('---------------------')
```

## Inspect the puzzle

Rows and columns are numbered 1 to 9. Empty CSV fields become zeros in the table and underscores in the display. A zero denotes an unfilled cell, not an allowed Sudoku digit.


```python
data = StringIO(
'''7,,,,2,,4,8,
2,,6,,,8,,,5
5,,,9,,,,,
,,,1,5,,,,
,2,,,,,,6,
,,,,6,7,,,
,,,,,6,,,3
6,,,5,,,1,,4
,9,3,,4,,,,7''')

start_sudoku = pd.read_csv(
    data, sep=',', index_col=False, header=None, names=list(range(1, 10))
).fillna(0).astype(int)
start_sudoku.index = start_sudoku.columns
assert start_sudoku.shape == (9, 9)
assert start_sudoku.isin(range(10)).all().all()
print_sudoku(start_sudoku)
```
