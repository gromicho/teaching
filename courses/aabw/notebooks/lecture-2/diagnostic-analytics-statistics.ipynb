# Diagnostic analytics: the significance of statistics


```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats
```

```python
url = 'https://raw.githubusercontent.com/gromicho/teaching/main/data/weight-height.csv'
df = pd.read_csv(url)
```

```python
df
```

The numbers above are in [imperial](https://www.interexchange.org/articles/career-training-usa/2012/05/24/imperial-vs-metric-system/), may we prefer [metric](https://www.interexchange.org/articles/career-training-usa/2012/05/24/imperial-vs-metric-system/) then its quite simple.

```python
df.Height = df.Height * 2.54
df.Weight = df.Weight * 0.45359237
```

```python
df
```

```python
df[df.Gender == 'Female'].describe()
```

```python
df[df.Gender == 'Male'].describe()
```

```python
sns.displot( data=df, hue='Gender', x='Height', fill=True )
```

```python
sns.displot( data=df, hue='Gender', x='Height', fill=True, kde=True )
```

```python
sns.displot( data=df, hue='Gender', x='Height', fill=True, kind='kde' )
```

```python
sns.displot( data=df, hue='Gender', x='Weight', fill=True, kind='kde' )
```

Are these variances the same? We can use [Levene's test](https://en.wikipedia.org/wiki/Levene%27s_test).

```python
stats.levene(df[df.Gender == 'Male'].Height,df[df.Gender == 'Female'].Height)
```

Such p-value renders the hipothesis to be so unlikely that we should reject it.

```python
stats.ttest_ind(df[df.Gender == 'Male'].Height,df[df.Gender == 'Female'].Height,equal_var=False)
```

Also the average height of a men and a women are significantly different as well.
