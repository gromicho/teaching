# Predicting the present and the future


Note that statistics (and machine learning as well) detect and learn from correlation. In order to detect true causality, you need domain knowledge!

<img src='https://pbs.twimg.com/media/EwOrq3vWYAMu43f?format=jpg&name=small'>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
url = 'https://raw.githubusercontent.com/gromicho/teaching/main/data/weight-height.csv'
df = pd.read_csv(url)
```

# Can we relate height and weight?

This also allows us to predict the present: given the height of a gentlemen, what should be the weight?

```python
from sklearn import linear_model

# Create linear regression object
regr = linear_model.LinearRegression()

# Train the model using the training sets
x = df.Height.values.reshape(-1,1)
y = df.Weight.values.reshape(-1,1)
regr.fit(x,y)

model_line = regr.predict(x)
plt.scatter(x, y,  color='black')
plt.plot(x, model_line, color='blue', linewidth=3)
plt.xticks(())
plt.yticks(())
plt.show()
```

# Can we predict the future?

Let us make use of [Facebook's prophet](https://facebook.github.io/prophet/), installing it automatically only when the current environment does not provide it.

```python
import prophet
```

```python
covid = pd.read_csv('https://raw.githubusercontent.com/CSSEGISandData/COVID-19/master/csse_covid_19_data/csse_covid_19_time_series/time_series_covid19_confirmed_global.csv')
```

```python
covid
```

```python
country = covid[ (covid['Country/Region']=='Netherlands') & (covid['Province/State'].isna()) ]
country = country[country.columns[4:]]
country
```

```python
country = country.T
country.columns = ['y']
```

```python
country.plot(figsize=(8,3),rot=90)
plt.show()
```

```python
country['ds'] = pd.to_datetime(country.index)
country = country.reset_index(drop=True)
```

```python
model = prophet.Prophet(yearly_seasonality=True,daily_seasonality=True)
# fit the model
model.fit(country)
```

```python
future = pd.DataFrame([ '2023-{:02d}'.format(m) for m in range(3,13)])
future.columns = ['ds']
future['ds']= pd.to_datetime(future['ds'])
```

```python
forecast = model.predict(future)
```

```python
model.plot(forecast)
plt.show()
```
