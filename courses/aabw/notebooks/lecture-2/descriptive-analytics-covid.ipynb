## A descriptive view of worldwide data using `pandas`

## In `python` you `import` what you need to use

```python
import pandas as pd
```

Pandas [dataframes](https://pandas.pydata.org/docs/reference/frame.html) make data easy to handle.

The data the is displayed at [the corona dashbord](https://coronavirus.jhu.edu/map.html) is maintained at [this repository](https://github.com/CSSEGISandData/COVID-19).

```python
covid = pd.read_csv('https://raw.githubusercontent.com/CSSEGISandData/COVID-19/master/csse_covid_19_data/csse_covid_19_time_series/time_series_covid19_confirmed_global.csv')
```

```python
covid
```

Let us check the latest reported date.

```python
last_day = covid.columns[-1]
```

```python
last_day
```

There are more rows than countries...

```python
covid.shape,len(set(covid['Country/Region']))
```

```python
covid[covid['Country/Region'] == 'China']
```

Let us group the data by country.
First, we use a smaller dataframe with the columns from the 4th to the end, and we index it on the values of the column `Country/Region`.

```python
data = covid[covid.columns[4:]]
data.index = covid['Country/Region']
```

Let us look at this more focused dataframe.

```python
data
```

```python
grouped = data.groupby(level=0, axis=0).sum()
```

```python
grouped
```

```python
grouped.T.plot()
```

```python
import matplotlib.pyplot as plt

grouped.T.plot(figsize=(13,8),rot=90)
plt.legend(ncol=6,loc='center left',bbox_to_anchor=(1, 0.5))
```

We need a new package, to be able to control the legend layout and position.
Let us also make our figure larger and rotate the labels on the horizontal axis.

We may ne more interested in the countries with the largest number of cases.

```python
grouped.nlargest(20,[last_day]).T.plot(figsize=(13,8),rot=90)
```
