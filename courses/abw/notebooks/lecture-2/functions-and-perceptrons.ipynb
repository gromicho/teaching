A short Lecture 2 example: a threshold function and an AND perceptron.


```python
def h(z):
  if z >= 0:
    return 1
  return 0
```

```python
def AND(a,b):
  return -3+2*a+2*b
```

```python
for a in [0,1]:
  for b in [0,1]:
    c = AND(a,b)
    print(f'{a} {b} {c:2} {h(c)}')
```
