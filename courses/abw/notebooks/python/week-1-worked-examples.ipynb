# Python Indentation
* What is Indentation?

Indentation refers to the spaces at the beginning of a line of code. In many programming languages, indentation is used to make the code more readable, but in Python, it’s a critical part of the syntax.

* Why is Indentation Important?

**In Python, indentation is used to define blocks of code. A block of code is a group of statements that should be executed together, such as the body of a function, the contents of a loop, or the statements inside an if condition.** Unlike some other programming languages that use curly braces {} (C, C++, Java, and JavaScript) or keywords (Ruby and VB.NET) to define blocks, Python uses indentation to achieve the same thing.

## Python if Statement
An if statement executes a block of code only when the specified condition is met.


Syntax template (for reference):

```python
if condition:
  # body of if statement
```

Here, `condition` is a boolean expression, such as number > 5, that evaluates to either True or False.

* If `condition` evaluates to True, the body of the if statement is executed.
* If `condition` evaluates to False, the body of the if statement will be skipped from execution.

````python
num = 3
if num > 0:
    print(num, "is a positive number.") #body of if statement

print("This is always printed.")
````

````python
num = -1
if num > 0:
    print(num, "is a positive number.") #body of if statement

print("This is also always printed.")
````

## Python if...else Statement
An if statement can have an optional else clause. The else statement executes if the condition in the if statement evaluates to False.

Syntax template (for reference):

```python
if condition:
    # body of if statement
else:
    # body of else statement
```

Here, if the `condition` inside the if statement evaluates to

* True - the body of if executes, and the body of else is skipped.
* False - the body of else executes, and the body of if is skipped

````python
num = 3

if num >= 0:
    print("Positive or Zero") # body of if statement
else:
    print("Negative number") # body of if statement
````

## Python if…elif…else Statement
The if...else statement is used to execute a block of code among two alternatives.

However, if we need to make a choice between more than two alternatives, we use the if...elif...else statement.

Syntax template (for reference):

```python

elif condition2:

else:
    # body of else statement
```

Here, if the`condition1` inside the if statement evaluates to

* False -  then, if the `condition2` inside the elif statement evaluates to

````python
'''In this program,
we check if the number is positive or
negative or zero and
display an appropriate message'''

num = 3.4

if num > 0:
    print("Positive number")   # body of if
elif num == 0:
    print("Zero")  # body of elif
else:
    print("Negative number") # body of else
````

## for loop
The for loop in Python is a statement used to iterate over a sequence of elements, such as a list, tuple, string, or range. It allows you to execute a block of code multiple times, once for each item in the sequence. The body of for loop is separated from the rest of the code using indentation.

Python provides control statements to alter the flow of a for loop:

* break: Exits the loop prematurely.
* continue: Skips the rest of the current loop iteration and moves to the next iteration.
* else: Executes a block of code once the loop completes normally (i.e., not exited via break).

````python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
````

````python
for i in range(5):
    if i == 3:
        break
    print(i)
````

````python
for i in range(5):
    if i == 3:
        continue
    print(i)
````

````python
for i in range(5):
    print(i)
else:
    print("Loop finished")

````

## range function


The range() function generates a sequence of numbers, which you can then iterate over or convert into a list. It’s particularly useful when you want to repeat an action a specific number of times.


The range() function has three parameters:

range(start, stop, step)

* start: The number to start the sequence from. This is optional and defaults to 0 if not provided.

* stop: The number to stop the sequence before. The sequence will include numbers up to, **but not including**,  this value.

* step: The difference between each number in the sequence. This is optional and defaults to 1 if not provided.

The step value can also be negative, which allows you to generate a sequence of numbers in reverse.

````python
for i in range(2,11,2):
  print(i)
````

````python
for i in range(11,2,-2):
  print(i)
````

````python
for i in range(10,1,-2):
  print(i)
````

Although range() itself doesn’t create a list in memory (it generates the numbers on the fly, which is more memory-efficient), you can convert it to a list if you need to.

````python
r = range(5)
print(r)
print(type(r))
````

````python
numbers = list(range(5))
print(numbers)
print(type(numbers))
````

## list comprehension
List comprehension in Python is a concise way to create lists. It’s a powerful feature that allows you to generate a new list by applying an expression to each item in an existing list or any iterable, and it can optionally include conditions.
Instead of writing a loop to build a list, you can use a single line of code.

Syntax template (for reference):

```python
[expression for item in iterable if condition]
```

````python
words = ["Analytics", "for", "a", "Better", "World"]
lengths = []
for word in words:
    lengths.append(len(word))

print(lengths)
assert lengths == [len(word) for word in words]
````

````python
numbers = [1, 2, 3, 4, 5]
squared_even_numbers = []

for number in numbers:
    if number % 2 == 0:
        squared_even_numbers.append(number ** 2)

print(squared_even_numbers)
````

````python
squared_even_numbers = [number ** 2 for number in range(1,6) if number % 2 == 0]

print(squared_even_numbers)
````

# Exercise 1.9 — Why the loop does not stop

The original loop tests whether repeated addition of `0.1` reaches exactly `2`:

```python
x = 0
while x != 2:
    print(x)
    x += 0.1
```

Do not run that unbounded loop. Most decimal fractions cannot be represented exactly by binary floating-point numbers. First predict what twenty additions will produce. The following bounded experiment safely demonstrates the original problem.

````python
x = 0.0
steps = 0
while x != 2 and steps < 25:
    print(f"{steps:2d}: {x!r}")
    x += 0.1
    steps += 1

print(f"Stopped after {steps} steps; x == 2 is {x == 2}.")
assert steps == 25 and x != 2
````

Floating-point arithmetic approximates real arithmetic. For example, `0.1 + 0.2 == 0.3` is false with Python floats. Repeated addition in the original loop skips exact equality with `2`, so that equality condition cannot provide a reliable stopping rule.

For a fixed number of steps, use an integer counter and compute the value from it. If you need an approximate numerical comparison, use `math.isclose` with tolerances appropriate to the scale and purpose of your calculation. A tolerance does not replace a bound on an iterative algorithm.

````python
import math

values = [step / 10 for step in range(21)]
print(values)
assert len(values) == 21 and values[-1] == 2.0

accumulated = sum([0.1] * 20)
print(f"Twenty additions: {accumulated!r}")
print(f"Exact equality: {accumulated == 2.0}")
print(f"Close enough for this example: {math.isclose(accumulated, 2.0, rel_tol=1e-12)}")
assert math.isclose(accumulated, 2.0, rel_tol=1e-12)
````

## Slicing

Slices select part of a list, tuple or string. Omitted bounds cover the sequence in the direction of the step. With a positive step this means beginning to end; with a negative step this means end to beginning. A slice can safely extend past the sequence bounds, while an out-of-range single index raises `IndexError`.

Syntax template (for reference):

```python
mylist[start:stop:step] # items from the start position until the one before the stop position, taken by step
```

Syntax template (for reference):

```python
mylist[start:stop]  # default of step is 1
mylist[start:]      # default of stop is the position after the last item in the sequence
mylist[:stop]       # default of start is the start of the sequence
mylist[:]           # a copy of the whole sequence
```

````python
mylist= ["a", "b", "c", "d", "e", "f"]
````

````python
mylist[1:5]
````

````python
mylist[-1:]    # from -1 to end
````

````python
mylist[-2:]   # from -2 to end
````

````python
mylist[:-2]   # from 0 to -2
````

With a negative step, traversal runs backwards. An omitted start begins at the final element, and an omitted stop continues through index 0. An explicit stop of `0` excludes index 0; an explicit stop of `-1` refers to the final element, so it is different from an omitted stop.

````python
mylist[::-1]    # from end to start of the sequence
````

````python
mylist[2:0:-1]   # from 2 to 1
````

````python
mylist[:-3:-1]  # from end to -3
````

````python
mylist[-3::-1]  # from -3 to start
````
