## Functions

We can also put the code into a [function](https://docs.python.org/3/tutorial/controlflow.html#defining-functions). A function is a group of related statements that perform a specific task. Using functions, it is easy to use your code for different variables. Consider the examples below:

```python
# This function greets the name passed as a parameter
def greet(name):
  print( "Hello " + name + ". Good morning!")

greet('Sam')
```

```python
def perimeter_field(length, width):
    perimeter = length * 2 + width * 2
    return perimeter

perimeter_field(2, 5)
```

## If-statement

Use `=` for assignment and `==` for an equality comparison. Comparisons
produce Boolean values that can control an `if` statement. Predict each
comparison in the next cell before running it.


```python
x = 5
print(x == 1) # This prints out False
print(x != 1) # This prints out True
print(x < 7)  # This prints out True
```

To ease the above statements, an [if-statement](https://docs.python.org/3/tutorial/controlflow.html#if-statements) can be used.

```python
if x == 5:
  print("x is equal to 5")

if x < 2 or x > 4:
  print("x is smaller than 2 or larger than 4")

y = 3
if x == 5 and y == 3:
  print('x is equal to 5 and y is equal to 3')
```

Additionally, it is possible to add an else-statement.

```python
x = 2
if x == 2:
  print('x equals 2')
else:
  print('x does not equal 2')
```

## Collections: lists and dictionaries


Collections hold multiple objects. Lists keep a sequence of items;
dictionaries associate keys with values. Choosing the right structure
helps keep the meaning of your data clear.


### Lists

A [list](https://docs.python.org/3/tutorial/introduction.html#lists) in Python is one of the most common data structures in Python. A list can be described as an ordered and mutable Python container.

A Python list can be created the following ways:

```python
my_list = []
my_list = list()
```

Or by directly adding elements to a list, such as strings, integers, objects or a mixture of types.

```python
my_list = [2025, year, next_year, 'December']
print(my_list)
```

By using `.append()`, it is possible to add something to the list. There are lots of [methods](https://docs.python.org/3/tutorial/introduction.html#lists) for lists.

```python
my_list.append('January')
print(my_list)
```

Two lists can be combined very easily.
Lists can be also be sorted.

```python
list1 =[22, 2, 36]
list2 = [48, 15, 96]
list3 = list1 + list2
print(list3)

list3.sort()
print(list3)
```

Strings and lists both support indexing and slicing, but a string is
**not** a list. Strings are immutable sequences of characters; lists are
mutable sequences that can contain different kinds of objects.
See [Python's introduction to sequences](https://docs.python.org/3/tutorial/introduction.html).


```python
print(list3)

# We can call for the first item of this list
print(list3[0])

# We can call for the last item of this list
print( list3[-1])

# We can leave out the 3rd item of the list (with index 2) as follows:
print(list3[:2] + list3[3:])
```

### List comprehension

A [list comprehension](https://docs.python.org/3/tutorial/datastructures.html#list-comprehensions)
builds a list from an iterable, optionally filtering its elements. The
next cell expresses the same calculation with a loop and a comprehension.
Prefer the version you can explain clearly; a comprehension is not
automatically faster or better for every task.


```python
squares = []
for i in range(5):
    squares.append(i*i)
print(squares)

#Instead of the above, we only need one line to get the same list.
squares = [i*i for i in range(5)]
print(squares)
```

## Loops

A loop is used to execute the same code multiple times within a Python script. There are two types of loops in Python, for and while.

### For loop

A [for loop](https://docs.python.org/3/tutorial/controlflow.html#for-statements) will iterate over a given sequence.

```python
# Create a list
my_list = [1, 2, 4, 6, 7]

# Print every value of the list
for i in my_list:
    print(i)
```

```python
# Print numbers 0, 1, 2, 3, 4. We can do this easily using range(). Note that the range funtion is zero based.
for i in range(5):
    print(i)
```

### While-loop

A `while` loop repeats **while its condition is True**. The condition is
checked before each iteration, so the body may run zero times. Make sure
the loop can eventually stop. What would happen below if `count += 1`
were removed?


```python
# Prints out 0,1,2,3,4
count = 0
while count < 5:
    print(count)
    count += 1 # This is the same as count = count + 1
```

## Further examples retained from the original notebook

The next two functions are **worked examples**, not new answers to submit.
First predict their behaviour, then explain the loop bounds. For
`count_code`, the third character can vary: the function counts patterns
such as `code` and `cope`. For `has22`, the two `2`s must be adjacent.


```python
def count_code(text):
    count = 0
    for i in range(len(text) - 3):
        c = text[i]
        o = text[i + 1]
        e = text[i + 3]
        if c == 'c' and o == 'o' and e == 'e':
            count += 1
    return count
```

```python
result = count_code('codexxcode')
print(result)
```

```python
def has22(nums):
  for i in range(len(nums)-1):
    if nums[i] == 2 and nums[i+1] == 2:
      return True
  return False
```

```python
has22([1, 2, 1, 2])
```

```python
has22([1, 2, 2])
```

```python
# Check the worked examples on ordinary and boundary cases.
assert count_code('codexxcode') == 2
assert count_code('cope') == 1
assert count_code('') == 0
assert count_code('co') == 0
assert has22([1, 2, 2]) is True
assert has22([2, 1, 2]) is False
assert has22([]) is False
assert has22([2]) is False
print('Worked-example checks passed. Explain why each expected result is correct.')
```
