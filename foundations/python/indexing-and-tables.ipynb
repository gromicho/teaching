## Names, indices and units
A mathematical subscript does not automatically start at zero. Use named dictionaries to make units and associations explicit, then compare them with a matrix representation.


```python
# Use installed packages, install only missing ones, without version pins.
from importlib.util import find_spec
import subprocess
import sys

required_packages = {
    'numpy': 'numpy',
    'pandas': 'pandas',
}
missing_packages = [
    package_name
    for import_name, package_name in required_packages.items()
    if find_spec(import_name) is None
]
if missing_packages:
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-q', *missing_packages])
```

```python
products = ['football','golf']
profit = {'football':12, 'golf':9}
wood = {'football':4, 'golf':2}
plan = {'football':650, 'golf':1100}
print(sum(profit[p]*plan[p] for p in products))
assert sum(wood[p]*plan[p] for p in products) == 4800
```

```python
import pandas as pd
import numpy as np
table = pd.DataFrame({'profit':profit,'wood':wood})
display(table)
print(table.loc['football','wood'])
print(table.iloc[0,1])
assert table.loc['golf','profit'] == 9
```

## Align before multiplying
Array multiplication uses position; pandas operations often align labels. Reordering a table should not change the meaning of a named calculation.


```python
quantities = pd.Series(plan)
reordered = table.iloc[::-1]
assert (reordered['profit']*quantities).sum() == 17700
A = table.loc[products,['wood']].to_numpy().T
x = quantities.loc[products].to_numpy()
assert float((A @ x)[0]) == 4800
```
