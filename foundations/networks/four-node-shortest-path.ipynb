## A simple shortest-path example

The network has four nodes, A, B, C and D. Direction matters: an edge from A to B does not automatically create an edge from B to A. Each edge has a `length` attribute; we must ask NetworkX to use it when finding the shortest path.


# Define a simple directed, weighted graph

```python
import matplotlib.pyplot as plt
import networkx as nx

g = nx.DiGraph()
g.add_edge('A','B',length= 1)
g.add_edge('A','C',length= 3)
g.add_edge('A','D',length= 4)
g.add_edge('B','C',length= 1)
g.add_edge('C','D',length= 1)
```

# Visualize it
`networkx` includes visualization possibilities, for details consult [the official documentation](https://networkx.org/documentation/stable/reference/drawing.html).


```python
pos = nx.circular_layout(g)
nx.draw(g, pos, with_labels=True)
labels = nx.get_edge_attributes(g,'length')
_= nx.draw_networkx_edge_labels(g, pos, edge_labels=labels)
# plt.savefig('SimpleGraph.pdf', bbox_inches='tight', pad_inches=0)
```

## Compute the shortest path

First find a shortest path from A to **C**, using `length` as the edge weight. A shortest path minimizes the sum of those weights, which is different from minimizing the number of edges. See the [NetworkX documentation](https://networkx.org/documentation/stable/reference/algorithms/shortest_paths.html).


```python
path = nx.shortest_path(g,source='A',target='C',weight='length')
path
```

## Turn the path into consecutive edges

Pair each node with the next one using `zip(path, path[1:])`. We can then look up and sum the lengths of those edges.


```python
path_edges = list(zip(path,path[1:]))
path_edges
```

The built-in `itertools` module provides another readable way to form consecutive pairs: `pairwise(path)`. Both approaches describe the same edges.


```python
from itertools import pairwise
list(pairwise(path))
```

```python
sum(g[u][v]['length'] for u, v in path_edges)
```

## Change the destination

The previous sum is the length from A to C. The final call below asks for the shortest-path length from A to **D**. Predict both values and identify the edges responsible for the difference.


```python
nx.shortest_path_length(g,source='A',target='D',weight='length')
```
